```python
import os, requests
import pandas as pd
from sqlalchemy import create_engine, text, Date
from sqlalchemy.engine import URL
pd.set_option("display.max_columns", 20)


def page_url(base, page, page_size=100):
    """URL satu halaman — mendukung API statis ('{page}' di URL) maupun API server (?page=)."""
    if "{page}" in base:
        return base.format(page=page)
    return f"{base}?page={page}&page_size={page_size}"
```

## Data mentah (sudah disiapkan)

```python
def extract_sales(api_url, page_size=100):
    """Mengambil seluruh halaman data dari API."""
    rows, page = [], 1
    while True:
        resp = requests.get(page_url(api_url, page, page_size), timeout=30)
        resp.raise_for_status()                      # error jika status bukan 2xx
        payload = resp.json()
        rows.extend(payload["data"])
        print(f"Halaman {page}: {len(payload['data'])} baris")
        if not payload.get("has_next"):
            break
        page += 1
    return pd.DataFrame(rows)


def load_csv_fallback(filename):
    """Cadangan jika API gagal: baca CSV dari folder data/ (clone lokal) atau langsung dari GitHub."""
    for p in (filename, f"data/{filename}", f"../data/{filename}"):
        if os.path.exists(p):
            return pd.read_csv(p, dtype=str, keep_default_na=False)
    url = f"{RAW_BASE}/data/{filename}"
    print("Membaca cadangan dari GitHub:", url)
    return pd.read_csv(url, dtype=str, keep_default_na=False)
```

## Tugas
Lengkapi `transform_sales()`:
1. Hapus duplikat
2. Ubah `transaction_date` menjadi tanggal (ada format `YYYY-MM-DD` **dan** `DD/MM/YYYY`)
3. Rapikan `city` (hapus spasi, huruf kapital di awal kata)
4. `quantity` → integer; nilai kosong diisi **1**
5. `unit_price` → angka; nilai negatif dijadikan positif
6. Tambahkan `total_amount = quantity * unit_price`

```python
import pandas as pd
import datetime as dt

def transform_sales(df):
    df = df.copy()

    # 1. Hapus duplikat
    print("Jumlah baris sebelum:", len(df))
    print("Jumlah duplikat:", df.duplicated().sum())

    df = df.drop_duplicates().copy()
    print("Jumlah baris setelah:", len(df))

    # 2. Konversi tanggal: YYYY-MM-DD dan DD/MM/YYYY
    df["transaction_date"] = pd.to_datetime(
        df["transaction_date"],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

    # 3. Rapikan nama kota
    df["city"] = (
        df["city"]
        .astype("string")
        .str.strip()
        .str.title()
    )

    # 4. Quantity menjadi integer, nilai kosong diisi 1
    df["quantity"] = (
        pd.to_numeric(df["quantity"], errors="coerce")
        .fillna(1)
        .astype(int)
    )

    # 5. Unit price menjadi angka positif
    df["unit_price"] = pd.to_numeric(
        df["unit_price"], errors="coerce"
    ).abs()

    # 6. Hitung total amount
    df["total_amount"] = df["quantity"] * df["unit_price"]

    return df


# Jalankan fungsi
df_clean = transform_sales(df_clean)

# Validasi hasil
assert len(df_clean) == 300, (
    f"Jumlah baris {len(df_clean)}, seharusnya 300"
)

assert df_clean["transaction_date"].notna().all(), (
    "Ada tanggal yang gagal dikonversi"
)

assert all(
    isinstance(x, (dt.date, pd.Timestamp))
    for x in df_clean["transaction_date"]
), "Tanggal belum dikonversi"

assert df_clean["transaction_date"].max() <= pd.Timestamp("2026-03-31"), (
    "Ada tanggal tertukar hari/bulan atau di luar periode"
)

assert set(df_clean["city"]) == {
    "Malang", "Surabaya", "Batu", "Sidoarjo", "Pasuruan"
}, f"Kota: {set(df_clean['city'])}"

assert str(df_clean["quantity"].dtype).startswith("int"), (
    "quantity harus integer"
)

assert (df_clean["unit_price"] >= 0).all(), (
    "Masih ada harga negatif"
)

assert int(df_clean["total_amount"].sum()) == 146_350_000, (
    f"Total tidak tepat: {df_clean['total_amount'].sum()}"
)

print(
    "✅ Exercise 2 benar! Total revenue Q1 = Rp",
    f"{df_clean['total_amount'].sum():,.0f}"
)

display(df_clean.head())
```
